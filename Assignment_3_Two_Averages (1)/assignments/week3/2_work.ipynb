```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load the illustrative teaching price data.
prices = pd.read_csv("illustrative_teaching_prices.csv")
prices["date"] = pd.to_datetime(prices["date"])
prices.head()
```

## 1. Mean, median and first-to-last percentage change

The mean is calculated by adding the prices and dividing by the number of observations. The median is the middle value after sorting the prices. The percentage change compares the first and last observations.

```python
summary = (
    prices.sort_values("date")
    .groupby("item")["price"]
    .agg(mean="mean", median="median", first="first", last="last", count="count")
)
summary["first_to_last_change_%"] = (
    (summary["last"] - summary["first"]) / summary["first"] * 100
)
summary = summary.round(2)
summary
```

## 2. Mean and median with and without a flagged value

The previous Assignment 2 draft identified a suspicious Staple A value of 1250. The full original CSV was not included, so this section uses an illustrative set of comparison values and includes 1250 as the flagged observation. The other values are not claimed to be recovered from the original file.

```python
week2 = pd.read_csv("illustrative_week2_flagged_value.csv")
all_values = week2["price"]
verified_only = week2.loc[~week2["flagged"], "price"]

comparison = pd.DataFrame({
    "Mean": [all_values.mean(), verified_only.mean()],
    "Median": [all_values.median(), verified_only.median()],
}, index=["Including flagged value", "Excluding flagged value"]).round(2)

comparison
```

### Interpretation

Including the flagged value of 1250 increases the mean considerably because the mean uses every observation. The median is less affected because it depends on the middle position after sorting. Excluding the flagged value gives a better summary of the ordinary values in this illustrative example, but the flagged value should remain documented rather than silently deleted. It should only be excluded from the main analysis until it can be verified.
